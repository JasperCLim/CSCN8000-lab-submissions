```python
#create two matrices
e1 = np.array([[5,7,2],[4,5,6],[7,4,2]])
e2 = np.array([[4,2],[6,2],[4,2]])

#it is possible when the 3x3 is multiplied onto the 2x3.
e3 = np.matmul(e1,e2)
print(e3)

#it is not possible when the 2x3 is multiplied onto the 3x3. matrix multiplication is not commutative.
try:
    e4 = np.matmul(e2,e1)
except ValueError as msg:
    print(msg)
```

```python
# Broadcasting applies a 1D array to each row of a compatible 3x3 matrix.
matrix = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])
vector = np.array([1, 2, 3])

# The vector is broadcast across all three rows.
subtraction = matrix - vector
multiplication = matrix * vector

print("Matrix:")
print(matrix)
print("Vector broadcast across each row:")
print(vector)
print("Subtraction:")
print(subtraction)
print("Element-wise multiplication:")
print(multiplication)
```
